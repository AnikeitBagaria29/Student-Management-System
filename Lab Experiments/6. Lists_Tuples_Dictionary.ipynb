Question 1 Scan n values in range 0-3 and print number of times each value has occurred.

```python
given_values = [0, 1, 2, 3, 2, 1, 0, 3, 3, 2]
counts = {0: 0, 1: 0, 2: 0, 3: 0}

for value in given_values:
    if value in counts:
        counts[value] += 1

for key in counts:
    print(f"{key} occurred {counts[key]} time(s)")
```

Question 4. Create a dictionary of persons where key is name and value is city, then:
Display all names, Display all city names, Display student names with cities, Count number of students in each city

```python
students = {"Anikeit": "Mumbai", "Priya": "Delhi", "Mohan": "Chennai", "Kirti": "Kolkata",}

print("Names:", list(students.keys()))
print("City names:", list(students.values()))

print("Student and city details:")
for name, cityname in students.items():
    print(f"{name} --- {cityname}")

city_count = {}
for city in students.values():
    city_count[city] = city_count.get(city, 0) + 1

print("Count in each city:")
for city, count in city_count.items():
    print(f"{city}: {count}")
```

Question 5. Store details of movies and perform:
Print all movie details, Display movies released before 2020, Print movies that made a profit
Print movies by a particular director

```python
movies = [
    {"Name": "RRR", "Year": 2022, "Director": "Rajamouli", "Cost": 1000, "Collection": 5000},
    {"Name": "Baahubali", "Year": 2017, "Director": "Rajamouli", "Cost": 650, "Collection": 500},
    {"Name": "Dangal", "Year": 2016, "Director": "Nitesh Tiwari", "Cost": 380, "Collection": 400},
]

print("All movie details:")
for movie in movies:
    print(movie)

print("Movies released before 2020:")
for movie in movies:
    if movie["Year"] < 2020:
        print(movie["Name"])

print("Movies that made profit:")
for movie in movies:
    if movie["Collection"] > movie["Cost"]:
        print(movie["Name"])

director = "Rajamouli"
print(f"Movies directed by {director}:")
for movie in movies:
    if movie["Director"].lower() == director.lower():
        print(movie["Name"])
```
