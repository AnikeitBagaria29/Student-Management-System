Question 8. Find area of a triangle when lengths of all three sides are given (Heron's formula).

```python
import math

a, b, c = 5, 6, 8
s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))
#print("Area of Triangle:", area)
print(f"Area of Triangle with Sides {a}, {b} and {c} is {area}")
```

Question 12. Print truth table for bitwise operators &, | and ^.

```python
print("a b | a&b a|b a^b")
for a in (0, 1):
    for b in (0, 1):
        print(a, b, "|", a & b, " ", a | b, " ", a ^ b)
```

Question 13. Find left shift and right shift values of a given number.

```python
num = 20
shift = 3

print("Number:", num)
print("Left shift by 3:", num << shift)
print("Right shift by 3:", num >> shift)
```
