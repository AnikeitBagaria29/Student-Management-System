Question 1. Add a few names (one name per row) in name.txt, then:
Count total names, Count names starting with a vowel, Find the longest name

```python
names = ["Ananya", "Priya", "Aarti", "Divya", "Rohit", "Vijay", "Deepak", "Anil"]

with open("name.txt", "w", encoding="utf-8") as file:
    for name in names:
        file.write(name + "\n")

with open("name.txt", "r", encoding="utf-8") as file:
    names_from_file = [line.strip() for line in file if line.strip()]
#    print(names_from_file)

vowels = tuple("aeiouAEIOU")
count_names = len(names_from_file)
count_vowel_start = sum(1 for name in names_from_file if name.startswith(vowels))
longest_name = max(names_from_file, key=len) if names_from_file else None

print("Names in file:", names_from_file)
print("Count of names:", count_names)
print("Names starting with vowel:", count_vowel_start)
print("Longest name:", longest_name)
```

Question 2. Store integers in a file and then:
Find the maximum number, Find the average of all numbers, Count numbers greater than 100

```python
numbers = [55, 110, 39, 8, 236, 76, 90, 32, 78]

with open("number.txt", "w", encoding="utf-8") as file:
    for number in numbers:
        file.write(f"{number}"+"\n")

with open("number.txt", "r", encoding="utf-8") as file:
    value = [int(line.strip()) for line in file if line.strip()]

max_no = max(value)
avg_no = sum(value) / len(value)
more_than_100 = sum(1 for number in value if number > 100)

print("Numbers:", value)
print("Maximum number:", max_no)
print("Average of numbers:", round(avg_no, 2))
print("Count > 100:", more_than_100)
```

Question 3. Assume city.txt stores city details as: city_name population_in_million area_in_sq_km. Read the file and:
Display details of all cities, Display city names with population more than 20 million, Display sum of areas of all cities

```python
city_records = [
    ("Delhi", 35.5, 1484),
    ("Mumbai", 22.5, 603),
    ("Kolkata", 16.1, 206),
    ("Bengaluru", 15.5, 741),
    ("Chennai", 13.1, 426),
]

with open("city.txt", "w", encoding="utf-8") as file:
    for city, population, area in city_records:
        file.write(f"{city} {population} {area}"+"\n")

cities = []
with open("city.txt", "r", encoding="utf-8") as file:
    for line in file:
        city_split = line.split()
        if len(city_split) == 3:
            city = city_split[0]
            population_million = float(city_split[1])
            area_sqkm = float(city_split[2])
            cities.append((city, population_million, area_sqkm))

print("All city details:")
for city, population_million, area_sqkm in cities:
    print(f"{city}: population = {population_million} million, area = {area_sqkm} sq km")

high_population_cities = [city for city, population_million, _ in cities if population_million > 20]
total_area = sum(area for area_sqkm in cities)

print("Cities with population > 20 million:", high_population_cities)
print("Sum of all city areas:", round(total_area, 2), "sq km")
```

Question 4. Input N test cases of a b, perform integer division a/b, and handle exceptions for ZeroDivisionError and ValueError.

```python
raw_input_data = "3 1 0 2 4 3 1"

lines = raw_input_data.strip().splitlines()
n = int(lines[0])

for line in lines[1:n + 1]:
    a_text, b_text = line.split()
    try:
        result = int(a_text) // int(b_text)
        print(result)
    except (ZeroDivisionError, ValueError) as error:
        print("Error Code:", error)
```

Question 5. Create multiple suitable exceptions for a file handling program.
Explanation: We define custom exceptions to handle domain-specific file issues:
EmptyFileError for empty files
InvalidRecordError for badly formatted records

```python
class EmptyFileError(Exception):
    pass

class InvalidRecordError(Exception):
    pass

def student_marks(file_name):
    try:
        with open(file_name, "r", encoding="utf-8") as file:
            lines = [line.strip() for line in file if line.strip()]
    except FileNotFoundError as error:
        raise FileNotFoundError(f"{file_name} not found.") from error
    if not lines:
        raise EmptyFileError(f"{file_name} is empty.")
    marks = {}
    for line in lines:
        line_split = line.split(",")
        if len(line_split) != 2:
            raise InvalidRecordError(f"Invalid line format: '{line}'")
        name = line_split[0].strip()
        try:
            marks_line = float(line_split[1].strip())
        except ValueError as error:
            raise InvalidRecordError(f"Marks must be numeric: '{line}'") from error
        marks[name] = marks_line
    return marks

with open("students_valid.txt", "w", encoding="utf-8") as file:
    file.write("Aditi, 88"+"\n"+"Rohan,76")

with open("students_invalid.txt", "w", encoding="utf-8") as file:
    file.write("Aditi, -88"+"\n"+"Rohan, abc")

for candidate in ["missing.txt", "students_invalid.txt", "students_valid.txt"]:
    try:
        data = student_marks(candidate)
        print(f"{candidate} -> {data}")
    except (FileNotFoundError, EmptyFileError, InvalidRecordError) as error:
        print(f"{candidate} -> {error}")
```
