Question 1 Count and display the number of capital letters in a given string

```python
string = "Cyber Security"
count_capital = sum(1 for ch in string if ch.isupper())
print(f"Capital letters in {string} is {count_capital}")
```

Question 2. Count total number of vowels in a given string.

```python
string = "Cyber Security"
vowels = "aeiouAEIOU"
count_vowels = sum(1 for ch in string if ch in vowels)
print(f"Total vowels in {string} is {count_vowels}")
```

Question 3. Input a sentence and print words in separate lines.

```python
sentence = "Print Words in Separate Lines"

for word in sentence.split():
    print(word)
```

Question 4. Enter a string and substring, then print how many times substring occurs from left to right.

```python
main_string = "This is a long sentence to test is in this sentence"
sub_string = "sentence"

count = 0
start = 0
while True:
    location = main_string.find(sub_string, start)
    if location == -1:
        break
    count += 1
    start = location + 1

print("Number of Occurrences of substring is:", count)
```

Question 5. Count occurrences of each alphabet (case-insensitive) in a mixed-case string.

```python
string = "Abrakadabra"
count = {}

for character in string:
    if character.isalpha():
        key = character.upper()
        count[key] = count.get(key, 0) + 1

for key in sorted(count):
    print(f"{count[key]}{key}")
```

Question 6. Count number of unique words in a sentence using sets.

```python
string = "This is a long sentence to test unique words in this sentence"
words = string.lower().split()
unique_words = set(words)

print("Unique words:", unique_words)
print("Count of unique words:", len(unique_words))
```

Question 7. Create two sets of fruits and find:
Fruits in both sets, Fruits only in s1, Count of all fruits from both sets

```python
Set_1 = {"apple", "banana", "mango", "grapes"}
Set_2 = {"banana", "orange", "kiwi", "mango"}

print("Fruits in both sets:", Set_1 & Set_2)
print("Fruits only in set 1:", Set_1 - Set_2)
print("Count of all unique fruits:", len(Set_1 | Set_2))
```

Question 8. Take two sets and apply various set operations:

```python
Set_1 = {"Red", "yellow", "orange", "blue"}
Set_2 = {"violet", "blue", "purple"}

print("Union:", Set_1 | Set_2)
print("Intersection:", Set_1 & Set_2)
print("Difference (Set 1 - Set 2):", Set_1 - Set_2)
print("Difference (Set 2 - Set 1):", Set_2 - Set_1)
print("Symmetric Difference:", Set_1 ^ Set_2)
```
