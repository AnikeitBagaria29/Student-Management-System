Question 1. Find factorial of a given number using a loop

```python
x = 10
factorial = 1

for i in range(1, x + 1):
    factorial *= i

print(f"Factorial of {x} is {factorial}")
```

Question 2. Check whether the given number is an Armstrong number.

```python
x = 370
x_str = str(x)
power = len(x_str)
sum_of_powers = 0

for digit in x_str:
    sum_of_powers += int(digit) ** power

if sum_of_powers == x:
    print(x, "is an Armstrong number")
else:
    print(x, "is not an Armstrong number")
```

Question 3. Print Fibonacci series up to a given number of terms.

```python
terms = 10
a = 0 
b = 1

print("Fibonacci series (for 15 terms):")
for x in range(terms):
    print(a, end=" ")
    a = b
    b = a + b
print()
```

Question 4. Check whether a given number is prime or not.

```python
x = 17

if x < 2:
    print(x, "is not prime")
else:
    is_prime = True
    for a in range(2, int(x ** 0.5) + 1):
        if x % a == 0:
            is_prime = False
            break
    print(x, "is prime" if is_prime else "is not prime")
     
```

Question 5. Check whether a given number is palindrome or not.

```python
x = 4567654

if str(x) == str(x)[::-1]:
    print(x, "is a palindrome")
else:
    print(x, "is not a palindrome")
```

Question 6. Print sum of digits of a number.

```python
x = 4672
sum_digits = 0

for digit in str(x):
    sum_digits += int(digit)

print("Sum of digits:", sum_digits)
```

Question 7. Count and print all numbers divisible by 5 or 7 between 1 and 100.

```python
num = [n for n in range(1, 101) if n % 5 == 0 or n % 7 == 0]
print("Numbers:", num)
print("Count:", len(num))
```

Question 9 Print all prime numbers between 1 and 100.

```python
primes_num = []

for prime in range(2, 101):
    is_prime = True
    for a in range(2, int(prime ** 0.5) + 1):
        if prime % a == 0:
            is_prime = False
            break
    if is_prime:
        primes_num.append(prime)

print("Prime numbers from 1 to 100:", primes_num)
```
