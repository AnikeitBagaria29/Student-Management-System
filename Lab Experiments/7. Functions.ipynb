Question 1. Write a function to find maximum and minimum from a sequence without using built-in max() and min()

```python
def max_min(numbers):
    maximum_no = numbers[0]
    minimum_no = numbers[0]
    for value in numbers[1:]:
        if value > maximum_no:
            maximum_no = value
        if value < minimum_no:
            minimum_no = value
    return maximum_no, minimum_no

number_list = [45, 76, 34, 90, 12]
max, min = max_min(number_list)
print("Maximum Number in list is:", max)
print("Minimum Number in list is:", min)
```

Question 2. Write a function that takes a positive integer and returns sum of cubes of all positive integers smaller than it.

```python
def sum_of_cubes(c):
    sum = 0
    for i in range(1, c):
        sum += i ** 3
    return sum

x = 8
cubes = sum_of_cubes(x)
print(f"Sum of cubes below {x}:", cubes)
```

Question 3. Write a recursive function to print numbers from 1 to n (no loops).

```python
def print_series(n):
    if n == 0:
        return
    print_series(n - 1)
    print(n, end=" ")

print_series(10)
```

Question 4. Write a recursive function to print Fibonacci series up to n terms.

```python
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    fib = fibonacci(n - 1) + fibonacci(n - 2)
    return fib

count = 10
print("Fibonacci series:")
for i in range(count):
    print(fibonacci(i), end=" ")
```

Question 5. Write a lambda function to find volume of a cone.
V = (1/3) * pi * r^2 * h

```python
import math
volume = lambda r, h: (1 / 3) * math.pi * r ** 2 * h
print("Volume of cone:", volume(4, 7))
```

Question 7. Write functions to explain: Keyword arguments, Default arguments, Variable length arguments

```python
#Keyword arguments
def student(name, age):
    print(f"Name: {name}, Age: {age}")
student(age=23, name="Anikeit")

#Default arguments
def greet(name, message="Hello"):
    print(f"{name}: {message}")
greet("Anikeit")
greet("Anikeit", "How are you?")

# Variable length argument
def total_marks(*marks):
    return sum(marks)
print("Total marks:", total_marks(82, 89, 95, 92, 78))
```
