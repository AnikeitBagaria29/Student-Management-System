Question 1. Check whether a given number is divisible by both 3 and 5.

```python
num = 75

if num % 3 == 0 and num % 5 == 0:
    print(num, "is divisible by both 3 and 5")
else:
    print(num, "is not divisible by both 3 and 5")

num = 100

if num % 3 == 0 and num % 5 == 0:
    print(num, "is divisible by both 3 and 5")
else:
    print(num, "is not divisible by both 3 and 5")
```

Question 5. Check whether a quadratic equation has real or imaginary roots and display the roots.
Equation: ax^2 + bx + c = 0

```python
import cmath

a, b, c = 9, 5, 4

discriminant = b ** 2 - 4 * a * c
root1 = (-b + cmath.sqrt(discriminant)) / (2 * a)
root2 = (-b - cmath.sqrt(discriminant)) / (2 * a)

if discriminant >= 0:
    print("Roots are real")
else:
    print("Roots are imaginary")

print("Root 1:", root1)
print("Root 2:", root2)
```

Question 6. Find whether a given year is a leap year or not.

```python
year = 2026

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

Question 7. Take any date as input and display the next date in the calendar

```python
from datetime import date, timedelta

input_day = date(2026, 7, 23)
next_day = input_day + timedelta(days=1)

print("Input Date:", input_day.strftime("%d-%m-%Y"))
print("Next Date:", next_day.strftime("%d-%m-%Y"))
```

Question 8. Print the grade sheet of a student from marks of five subjects.
Percentage = (total marks / 500) * 100
CGPA = Percentage / 10
Grade is assigned using the provided CGPA ranges.

```python
def get_grade(cgpa):
    if 0.0 <= cgpa <= 3.4:
        return "F"
    if 3.5 <= cgpa <= 5.0:
        return "C+"
    if 5.1 <= cgpa <= 6.0:
        return "B"
    if 6.1 <= cgpa <= 7.0:
        return "B+"
    if 7.1 <= cgpa <= 8.0:
        return "A"
    if 8.1 <= cgpa <= 9.0:
        return "A+"
    if 9.1 <= cgpa <= 10.0:
        return "O (Outstanding)"
    return "Invalid CGPA"

name = "Anikeit Bagaria"
roll_no = "R17234512"
sap_id = "590040405"
semester = 1
course = "M.Tech. CSE Data Science"

marks = {
    "Data Science": 75,
    "Python": 85,
    "AI ML": 90,
    "Java": 65,
    "Maths": 70,
}

total = sum(marks.values())
percentage = (total / (len(marks) * 100)) * 100
cgpa = round(percentage / 10, 1)
grade = get_grade(cgpa)

print("Sample Gradesheet")
print("Name:", name)
print("Roll Number:", roll_no, "SAPID:", sap_id)
print("Semester:", semester, "Course:", course)
print("Subject name: Marks")
for subject, score in marks.items():
    print(f"{subject}: {score}")
print(f"Percentage: {percentage:.2f}%")
print("CGPA:", cgpa)
print("Grade:", grade)
```
